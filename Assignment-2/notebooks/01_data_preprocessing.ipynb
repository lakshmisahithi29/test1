```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Pandas version:", pd.__version__)
```

```python
df = pd.read_csv("../data/adult.csv")

df.head()
```

```python
df.replace("?", np.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df.drop(columns=["fnlwgt"], inplace=True)
df.dtypes
```

```python
num_cols = df.select_dtypes(include=["int64", "float64"]).columns
cat_cols = df.select_dtypes(include="object").columns

num_cols, cat_cols
```

```python
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.duplicated().sum()
```

```python
Q1 = df[num_cols].quantile(0.25)
Q3 = df[num_cols].quantile(0.75)
IQR = Q3 - Q1

df = df[~((df[num_cols] < (Q1 - 1.5 * IQR)) | 
          (df[num_cols] > (Q3 + 1.5 * IQR))).any(axis=1)]
```

```python
# Data cleaning summary:
# Categorical missing values were handled using mode imputation

# Numerical columns had no missing values

# fnlwgt was removed due to lack of predictive relevance

# IQR-based outlier removal was applied to numerical feature
```

```python
# Categorical Encoding Summary
# Different categorical encoding techniques were applied to demonstrate their suitability for various types of categorical features.

# Label Encoding was used for quick transformation of categorical values into numeric form, but it was noted that it may introduce unintended ordinal relationships for nominal variables.

# One-Hot Encoding was applied to nominal features such as workclass, occupation, race, and sex, as it prevents the model from assuming any inherent order among categories.

# Ordinal Encoding was specifically used for the education feature because it has a natural hierarchical order ranging from lower to higher educational qualifications.

# Frequency Encoding was used to replace categories with their occurrence frequencies, which helps retain information about category distribution without increasing dimensionality.

# Target Encoding was applied cautiously, as it uses the target variable to encode categories and may introduce data leakage if not handled properly. It was demonstrated only for educational purposes.

# All major encoding techniques were implemented to satisfy the assignment requirements and to compare their applicability across different types of categorical variables.
```
