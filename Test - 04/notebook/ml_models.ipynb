```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../dataset/adult.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
# # Target variable 
# The target variable for this classification task is income, which indicates whether an individual earns more than 50K per year.
```

```python
df['income'].value_counts()
```

```python
# ### Initial Data Observations
# Number of rows and columns

# Mix of categorical and numerical features

# Presence of missing-value placeholders (?)

# Binary target variable

# Need for encoding and scaling
```

```python
df.replace("?", np.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
num_cols = df.select_dtypes(include=["int64", "float64"]).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
cat_cols = df.select_dtypes(include=["object", "string"]).columns


for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.drop(columns=["fnlwgt"], inplace=True, errors="ignore")
```

```python
num_cols = df.select_dtypes(include=["int64", "float64"]).columns
```

```python
num_cols = [col for col in num_cols if col != 'income']
```

```python
Q1 = df[num_cols].quantile(0.25)
Q3 = df[num_cols].quantile(0.75)
IQR = Q3 - Q1

df = df[~((df[num_cols] < (Q1 - 1.5 * IQR)) | 
          (df[num_cols] > (Q3 + 1.5 * IQR))).any(axis=1)]
```

```python
df.shape
```

```python
# ### Data Cleaning Summary
# Missing values were handled using median imputation for numerical features and mode imputation for categorical features. 
# Duplicate records were removed to prevent biased learning. The fnlwgt feature was dropped due to its lack of predictive relevance. 
# Outliers were treated using the IQR method to improve model stability.
```

```python
X = df.drop(columns=['income'])
y = df['income']
```

```python
y = y.map({'<=50K': 0, '>50K': 1})
```

```python
y.value_counts()
```

```python
X = pd.get_dummies(X, drop_first=True)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
# ### Encoding, Scaling and Data Split
# Categorical features were converted into numerical form using one-hot encoding to make them suitable for machine learning algorithms.
#  Feature scaling was applied using standardization to ensure equal contribution of all features, which is especially important for distance-based models such as KNN and SVM.
#  The dataset was split into training and testing sets using an 80–20 ratio.
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
def evaluate_model(name, y_true, y_pred):
    print(f"--- {name} ---")
    print("Accuracy :", accuracy_score(y_true, y_pred))
    print("Precision:", precision_score(y_true, y_pred))
    print("Recall   :", recall_score(y_true, y_pred))
    print("F1-score :", f1_score(y_true, y_pred))
    print()
```

```python
#Model 1 :LOGISTIC REGRESSION(Linear Model)
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train, y_train)

y_pred_lr = log_reg.predict(X_test)

evaluate_model("Logistic Regression", y_test, y_pred_lr)
```

```python
#Model 2: DECISION TREE
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)

evaluate_model("Decision Tree", y_test, y_pred_dt)
```

```python
#MODEL 3: Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

evaluate_model("Random Forest", y_test, y_pred_rf)
```

```python
#MODEL 4: K-Nearest Neighbors (KNN)
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

y_pred_knn = knn.predict(X_test)

evaluate_model("KNN", y_test, y_pred_knn)
```

```python
#MODEL 5: Support Vector Machine (SVM)
from sklearn.svm import SVC

svm = SVC(kernel='rbf')
svm.fit(X_train, y_train)

y_pred_svm = svm.predict(X_test)

evaluate_model("SVM", y_test, y_pred_svm)
```

```python
# ### Model Training and Evaluation
# Five supervised machine learning algorithms were trained and evaluated on the Adult Census dataset. 
# Logistic Regression served as a linear baseline model. 
# Tree-based models such as Decision Tree and Random Forest captured non-linear relationships. 
# Distance-based models like KNN and SVM benefited from feature scaling. 
# Model performance was compared using accuracy, precision, recall, and F1-score.
```

```python
results = {
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "KNN",
        "SVM"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_knn),
        accuracy_score(y_test, y_pred_svm)
    ],
    "Precision": [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_dt),
        precision_score(y_test, y_pred_rf),
        precision_score(y_test, y_pred_knn),
        precision_score(y_test, y_pred_svm)
    ],
    "Recall": [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_dt),
        recall_score(y_test, y_pred_rf),
        recall_score(y_test, y_pred_knn),
        recall_score(y_test, y_pred_svm)
    ],
    "F1-Score": [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_dt),
        f1_score(y_test, y_pred_rf),
        f1_score(y_test, y_pred_knn),
        f1_score(y_test, y_pred_svm)
    ]
}

results_df = pd.DataFrame(results)
results_df
```

```python
# ### Model Comparison
# The performance of five supervised machine learning models was compared using accuracy, precision, recall, and F1-score. 
# Random Forest generally achieved the best overall performance due to its ability to handle non-linear relationships and reduce overfitting through ensemble learning.
#  Logistic Regression provided a strong baseline, while KNN and SVM benefited from feature scaling.
#  Decision Tree showed good interpretability but was more prone to overfitting.
```
